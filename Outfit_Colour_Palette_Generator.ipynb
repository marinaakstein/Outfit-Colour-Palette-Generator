```python
# import random so the program can randomly choose colours
import random

# matplotlib lets us display the colours visually
import matplotlib.pyplot as plt


# create the different fashion colour themes
themes = {

    "Y2K": [
        ("Hot Pink", "#FF4FA3"),
        ("Aqua", "#63D9E6"),
        ("Lavender", "#B99CFF"),
        ("Silver", "#C7C9D3"),
        ("Bubblegum Pink", "#FF9BCB"),
        ("Lime", "#B8F45D"),
        ("White", "#F7F7F7")
    ],

    "Earthy": [
        ("Olive", "#7D8448"),
        ("Terracotta", "#C96F4A"),
        ("Sand", "#D8C3A5"),
        ("Brown", "#5B3A29"),
        ("Sage", "#A8B79A"),
        ("Mustard", "#C99A2E"),
        ("Cream", "#F0E2C6")
    ],

    "Bold": [
        ("Red", "#D62828"),
        ("Blue", "#2554C7"),
        ("Black", "#171717"),
        ("Gold", "#D4A017"),
        ("Green", "#16805B"),
        ("Purple", "#6C3BAA"),
        ("White", "#F6F6F6")
    ],

    "Grunge": [
        ("Charcoal", "#333333"),
        ("Oxblood", "#5A1F2B"),
        ("Moss", "#5E6640"),
        ("Denim", "#6F829A"),
        ("Stone", "#8B8178"),
        ("Rust", "#9A4F32"),
        ("Cream", "#D9D0C2")
    ],

    "Soft": [
        ("Blush", "#E9B7B7"),
        ("Powder Blue", "#B9D6EA"),
        ("Lilac", "#D6C2E9"),
        ("Butter Yellow", "#F4E4A6"),
        ("Mint", "#BFDCC8"),
        ("Peach", "#F2C1A6"),
        ("Cream", "#F6EFE5")
    ]
}
```

```python
# this function creates the outfit colour palette
def generate_palette(theme_name):

    # randomly choose 5 different colours from the selected theme
    palette = random.sample(themes[theme_name], 5)

    # create the area where the colour palette will be displayed
    fig, ax = plt.subplots(figsize=(10, 3.5))

    # go through each colour in the palette
    for i, (colour_name, colour_code) in enumerate(palette):

        # draw each colour as a rectangle
        ax.bar(i, 1, color=colour_code, width=1)

        # write the colour name underneath
        ax.text(
            i,
            -0.07,
            colour_name,
            ha="center",
            va="top",
            fontsize=10,
            fontweight="bold"
        )

        # write the hex code underneath the colour name
        ax.text(
            i,
            -0.16,
            colour_code,
            ha="center",
            va="top",
            fontsize=8
        )

    # title of the palette
    ax.set_title(
        theme_name + " Outfit Palette",
        fontsize=18,
        fontweight="bold"
    )

    # remove graph lines and numbers
    ax.set_xlim(-0.5, 4.5)
    ax.set_ylim(-0.4, 1)
    ax.axis("off")

    # show the finished palette
    plt.show()
```

```python
generate_palette("")
```

```python
# import widgets so we can make buttons and dropdown menus
import ipywidgets as widgets

# lets us display and update the palette area
from IPython.display import display, clear_output


# create a dropdown menu with all of the fashion themes
theme_dropdown = widgets.Dropdown(
    options=["Y2K", "Earthy", "Bold", "Grunge", "Soft"],
    value="Y2K",
    description="Choose a theme:"
)


# create the generate button
generate_button = widgets.Button(
    description="Generate Palette",
    button_style="success"
)


# create an area where the palette will appear
output = widgets.Output()


# this function runs when the button is clicked
def on_generate_clicked(button):

    # get the theme currently selected in the dropdown
    selected_theme = theme_dropdown.value

    # clear the previous palette
    with output:
        clear_output(wait=True)

        # generate a new palette using the selected theme
        generate_palette(selected_theme)


# connect the button to the function above
generate_button.on_click(on_generate_clicked)


# display the dropdown, button and output
display(theme_dropdown)
display(generate_button)
display(output)
```

```python
# import widgets so we can make buttons and dropdown menus
import ipywidgets as widgets

# lets us display and update the palette area
from IPython.display import display, clear_output


# create a dropdown menu with all fashion themes
theme_dropdown = widgets.Dropdown(
    options=["Surprise Me", "Y2K", "Earthy", "Bold", "Grunge", "Soft"],
    value="Surprise Me",
    description="Choose a theme:"
)


# create the generate button
generate_button = widgets.Button(
    description="Generate Palette",
    button_style="success"
)


# create an area where the palette will appear
output = widgets.Output()


# this function runs when the button is clicked
def on_generate_clicked(button):

    # get the theme currently selected
    selected_theme = theme_dropdown.value

    # if Surprise Me is selected, randomly choose a theme
    if selected_theme == "Surprise Me":
        selected_theme = random.choice(
            ["Y2K", "Earthy", "Bold", "Grunge", "Soft"]
        )

    # clear the previous palette
    with output:
        clear_output(wait=True)

        # generate the new palette
        generate_palette(selected_theme)


# connect the button to the function
generate_button.on_click(on_generate_clicked)


# display everything
display(theme_dropdown)
display(generate_button)
display(output)
```
